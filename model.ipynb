```python
import tensorflow as tf
import os
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt
import numpy as np
from sklearn.model_selection import train_test_split
from tensorflow import keras
```

```python
train_dir='C:/Users/DELL/Desktop/alzheimer disease detection using dl/OriginalDataset/train/' # change the path 
test_dir='C:/Users/DELL/Desktop/alzheimer disease detection using dl/OriginalDataset/test/'
```

```python
batch_size = 1
epochs = 5
img_height = 224
img_width = 224
```

```python
train_image_generator = ImageDataGenerator(rescale=1./255)  
train_data_gen = train_image_generator.flow_from_directory(batch_size=batch_size,directory=train_dir,shuffle=True,target_size=(img_height, img_width),class_mode='categorical')
```

```python
val_image_generator = ImageDataGenerator(rescale=1./255)  
val_data_gen = val_image_generator .flow_from_directory(batch_size=batch_size,directory=test_dir,shuffle=True,target_size=(img_height, img_width),class_mode='categorical')
```

```python
import warnings
 

import os
import glob
import matplotlib.pyplot as plt

# Import Keras
import keras
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Dropout,Flatten
from tensorflow.keras.layers import Conv2D,MaxPooling2D,Activation,AveragePooling2D,BatchNormalization
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
img_width,img_height =224,224
input_shape=(img_width,img_height,3)
```

```python
model = Sequential()
model.add(Conv2D(32, (5, 5),input_shape=input_shape,activation='relu'))
model.add(MaxPooling2D(pool_size=(3, 3)))
model.add(Conv2D(32, (3, 3),activation='relu'))
model.add(MaxPooling2D(pool_size=(2, 2)))
model.add(Conv2D(64, (3, 3),activation='relu'))
model.add(MaxPooling2D(pool_size=(2, 2)))   
model.add(Flatten())
model.add(Dense(512,activation='relu'))
model.add(Dropout(0.25))
model.add(Dense(128,activation='relu'))          
model.add(Dense(4,activation='softmax'))
```

```python
model.summary()
```

```python
model.compile(optimizer='adam',
loss='categorical_crossentropy',
metrics=['accuracy'])
```

```python
history = model.fit(train_data_gen, epochs=8,
validation_data= val_data_gen,)
```
